```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print(df.head())
```

```python
subjects = ["math score", "reading score", "writing score"]

average_scores = df[subjects].mean()

plt.figure(figsize=(8, 5))
plt.bar(average_scores.index, average_scores.values)

plt.title("Average Student Scores by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

### 1. Average Student Scores by Subject

**Observation:** This chart compares the average scores of students in mathematics, reading, and writing.

**Analysis:** By comparing the bars, we can identify which subject has the highest average score and which has the lowest average score.

**Insight:** This comparison helps us understand the overall performance of students across the three subjects.


```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="math score", bins=10, kde=True)

plt.title("Distribution of Mathematics Scores")
plt.xlabel("Mathematics Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

### 2. Distribution of Mathematics Scores

**Observation:** This histogram shows how mathematics scores are distributed among students.

**Analysis:** The bars represent the number of students who achieved scores within different score ranges. The distribution helps us understand where most students' scores are concentrated.

**Insight:** This chart helps identify the common mathematics score ranges and understand variations in students' performance.


```python
gender_scores = df.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean()

gender_scores.plot(kind="bar", figsize=(9, 5))

plt.title("Average Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.legend(title="Subject")

plt.tight_layout()
plt.show()
```

### 3. Average Scores by Gender

**Observation:** This bar chart compares the average mathematics, reading, and writing scores of male and female students.

**Analysis:** By comparing the bars, we can identify differences in average scores between male and female students across the three subjects.

**Insight:** This comparison helps us understand how academic performance varies by gender across mathematics, reading, and writing.


```python
score_columns = ["math score", "reading score", "writing score"]

correlation = df[score_columns].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Student Scores")

plt.tight_layout()
plt.show()
```

### 4. Correlation Between Student Scores

**Observation:** This heatmap shows the correlation between mathematics, reading, and writing scores.

**Analysis:** The correlation values range from -1 to +1. Values closer to +1 indicate a stronger positive relationship, while values closer to 0 indicate a weaker linear relationship.

**Insight:** This chart helps us understand how performance in one subject is related to performance in the other subjects.


```python
parent_education = df.groupby(
    "parental level of education"
)[["math score", "reading score", "writing score"]].mean()

parent_education.plot(kind="bar", figsize=(12, 6))

plt.title("Student Performance by Parental Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Score")
plt.ylim(0, 100)
plt.xticks(rotation=30, ha="right")
plt.legend(title="Subject")

plt.tight_layout()
plt.show()
```

### 5. Student Performance by Parental Education

**Observation:** This bar chart compares the average mathematics, reading, and writing scores of students based on their parents' education levels.

**Analysis:** By comparing the bars across different parental education categories, we can identify differences in average student scores.

**Insight:** This visualization helps us explore the relationship between parental education level and students' academic performance.
